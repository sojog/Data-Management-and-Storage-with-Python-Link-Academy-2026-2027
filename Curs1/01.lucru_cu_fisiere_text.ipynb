## Imi da eroare daca incerc sa citesc un fisier care nu exista

```python
with open("01.un_fisier_care_NU_EXISTA!!!!!.txt","r") as file_handler:
    file_handler.readline()
    print(file_handler.readline())
```

# Pentru a scrie intr-un fisier - modul este w

```python
with open("01.un_ALT_fisier_care_NU_EXISTA!!!!!.txt","w") as file_handler:
    file_handler.write("Acest NOU text va aparea in fisier")
```

# Modul w - face suprascriere

```python
with open("01.un_ALT_fisier_care_NU_EXISTA!!!!!.txt","w") as file_handler:
    file_handler.write("1234567890")
```

# Modul a - face adaugare

```python
with open("01.un_ALT_fisier_care_NU_EXISTA!!!!!.txt","a") as file_handler:
    file_handler.write("\nqwertyui")
```
