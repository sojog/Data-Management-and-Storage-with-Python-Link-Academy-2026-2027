```python
import json
import requests
```

```python
base_url = "https://country.io/" 
country_path = "names.json"
```

```python
country_url = base_url + country_path
country_url
```

```python
response = requests.get(country_url)
response.status_code
```

```python
type(response.json())
```

```python
response.json().values()
```

```python
response.json().keys()
```

```python
response.json().items()
```

```python
country_code_names =  list(response.json().items())
country_code_names
```

```python
dict_country_to_code = {}
for code, country in country_code_names:
    dict_country_to_code[country] = code

dict_country_to_code
```
