```python
import json
import requests
```

```python
continent_url = "https://country.io/continent.json"
```

```python
continent_response = requests.get(continent_url)
set(continent_response.json().values())
```

```python
continent_json = continent_response.json()
```

```python
european_countries = []
south_american_countries = []
for country_code in continent_json:
    if continent_json[country_code] == "EU":
        european_countries.append(country_code)
    elif continent_json[country_code] == "SA":
        south_american_countries.append(country_code)
```

```python
south_american_countries
```

```python
country_url = "https://country.io/names.json"

country_code_response = requests.get(country_url)
country_codes = country_code_response.json()
country_codes
```

```python
for code in south_american_countries:
   print(country_codes[code])
```

```python
for code in european_countries:
   print(country_codes[code])
```

```python
europeans = sorted([country_codes[code] for code in european_countries])
europeans, len(europeans)
```
