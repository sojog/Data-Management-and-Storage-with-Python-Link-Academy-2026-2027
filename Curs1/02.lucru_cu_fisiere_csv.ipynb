```python
with open("02.culori.csv") as file_handler:
    text = file_handler.read()

text
```

```python
with open("02.culori.csv") as file_handler:
    text = file_handler.readlines()

text
```

```python
with open("02.culori.csv") as file_handler:
    linii_text = file_handler.readlines()

valori = []
for linie in linii_text:
    valori.append(linie.split(", "))
valori
```

```python
with open("02.culori.csv") as file_handler:
    linii_text = file_handler.readlines()

valori = []
for linie in linii_text:
    valori.append(linie.replace("\n", "").split(", "))
valori
```

# Modul built-in in Python csv

```python
import csv
```

```python
with open("02.culori.csv") as file_handler:
    csv_reader = csv.DictReader(file_handler)
    print(csv_reader)
```

```python
with open("02.culori.csv") as file_handler:
    csv_reader = csv.DictReader(file_handler)
    print(csv_reader)

    for line in csv_reader:
        print(line)
```
